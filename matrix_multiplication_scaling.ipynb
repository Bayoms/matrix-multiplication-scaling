```python
import random
import time

n = 500
A = [[random.randint(0, 9) for _ in range(n)] for _ in range(n)]
B = [[random.randint(0, 9) for _ in range(n)] for _ in range(n)]
print("Matrices ready:", len(A), "x", len(A[0]))
```

```python
def matmul(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            total = 0
            for k in range(n):
                total += A[i][k] * B[k][j]
            C[i][j] = total
    return C
```

```python
start = time.time()
C = matmul(A, B)
end = time.time()
print("500 x 500 took", round(end - start, 2), "seconds")
```

```python
import numpy as np
print("Correct:", np.array_equal(np.array(C), np.array(A) @ np.array(B)))
```

```python
n = 1000
A2 = [[random.randint(0, 9) for _ in range(n)] for _ in range(n)]
B2 = [[random.randint(0, 9) for _ in range(n)] for _ in range(n)]
start = time.time()
C2 = matmul(A2, B2)
end = time.time()
print("1000 x 1000 took", round(end - start, 2), "seconds")
```

```python
start = time.time()
np.array(A2) @ np.array(B2)
print("NumPy 1000 x 1000 took", round(time.time() - start, 3), "seconds")
```
